Object-Oriented Programming (OOP) in Python

```python
class Person:
    # Constructor method: called automatically when an object is created
    def __init__(self, name, age):
        self.name = name  # attribute
        self.age = age    # attribute

    # Method (function inside class)
    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")

# Creating objects (instances)
person1 = Person("Alice", 25)
person2 = Person("Bob", 30)

# Calling method
person1.introduce()
person2.introduce()
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # private attribute

    def deposit(self, amount):
        self.__balance += amount
        print(f"Deposited {amount}. New balance: {self.__balance}")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
        else:
            print("Insufficient balance!")

    def get_balance(self):
        return self.__balance

# Create object
account = BankAccount("John", 1000)
account.deposit(500)
account.withdraw(200)

# Access private variable using method
print("Balance is:", account.get_balance())
```

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

# Dog inherits from Animal
class Dog(Animal):
    def speak(self):
        print("Dog barks")

# Cat inherits from Animal
class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Create objects
a = Animal()
d = Dog()
c = Cat()

a.speak()
d.speak()
c.speak()
```

```python
class Bird:
    def fly(self):
        print("Bird can fly")

class Airplane:
    def fly(self):
        print("Airplane is flying")

# Function that accepts any object with a fly() method
def lets_fly(entity):
    entity.fly()

b = Bird()
a = Airplane()

lets_fly(b)
lets_fly(a)
```

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_info(self):
        print(f"Brand: {self.brand}, Model: {self.model}")

class Car(Vehicle):  # Inherits from Vehicle
    def __init__(self, brand, model, fuel):
        super().__init__(brand, model)  # Call parent constructor
        self.__fuel = fuel  # Encapsulation (private attribute)

    def display_info(self):  # Polymorphism
        super().display_info()
        print(f"Fuel Type: {self.__fuel}")

car1 = Car("Toyota", "Innova", "Diesel")
car1.display_info()
```

Writing and Using Functions in Python

```python
def greet():
    print("Hello! Welcome to Python Functions.")

# Calling the function
greet()
```

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
greet_user("Bob")  # You can reuse the function with different arguments
```

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()         # Uses default
greet_user("Ravi")   # Overrides default
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

04 – Dictionaries & Sets

```python
student = {
    'name'   : 'Priya',     # key 'name' maps to value 'Priya'
    'age'    : 22,           # key 'age'  maps to integer 22
    'grade'  : 'A',          # key 'grade' maps to string 'A'
    'score'  : 94.5          # key 'score' maps to float 94.5
}

print(student)               # prints the full dictionary
print(student['name'])       # access value by key → 'Priya'
print(student['score'])      # access score → 94.5
```

```python
constants = {
    'pi'         : 3.14159265,   # ratio of circumference to diameter
    'e'          : 2.71828182,   # Euler's number (base of natural log)
    'phi'        : 1.61803398,   # golden ratio
    'sqrt2'      : 1.41421356,   # square root of 2
}

print('π  =', constants['pi'])        # 3.14159265
print('e  =', constants['e'])         # 2.71828182
print('φ  =', constants['phi'])       # 1.61803398
print('√2 =', constants['sqrt2'])     # 1.41421356
```

```python
marks = {'Math': 88, 'Physics': 75, 'Chemistry': 82}  # initial marks

marks['Biology'] = 91           # add a new subject
marks['Math']    = 95           # update existing value
del marks['Chemistry']          # remove a key

print('Updated marks:', marks)  # {'Math': 95, 'Physics': 75, 'Biology': 91}

# Iterating over key-value pairs
for subject, score in marks.items():       # .items() gives (key, value) pairs
    print(f'  {subject:10s}: {score}')     # formatted output
```

```python
A = {1, 2, 3, 4, 5}     # set A
B = {3, 4, 5, 6, 7}     # set B

print('A           :', A)
print('B           :', B)
print('Union A|B   :', A | B)   # all elements in A or B (no duplicates)
print('Intersection:', A & B)   # elements common to both A and B → {3,4,5}
print('Difference  :', A - B)   # elements in A but NOT in B → {1,2}
print('Symmetric Δ :', A ^ B)   # elements in A or B but NOT both → {1,2,6,7}
```

```python
numbers_with_duplicates = [1, 2, 2, 3, 3, 3, 4, 5, 5]  # raw list
unique_numbers = set(numbers_with_duplicates)            # convert to set

print('Original :', numbers_with_duplicates)   # has duplicates
print('Unique   :', unique_numbers)            # only unique values
```

05 – Conditional Logic & Mathematical Decisions

```python
# ── Simple if-else ────────────────────────────────────────────────────
number = -7   # the number to test

if number > 0:              # condition: is number positive?
    print('Positive number')     # runs if condition is True
elif number == 0:           # else-if: is number zero?
    print('Number is zero')      # runs if this condition is True
else:                       # none of the above matched
    print('Negative number')     # runs if all conditions are False
```

```python
# ── Piecewise mathematical function ──────────────────────────────────
# f(x) = x²  if x >= 0
#        -x  if x < 0
def f(x):
    if x >= 0:         # non-negative: use x squared
        return x ** 2
    else:              # negative: use absolute value
        return -x

# Test the function for different values
for val in [-3, -1, 0, 2, 5]:   # loop over test values
    print(f'f({val:3d}) = {f(val)}')
```

```python
# ── Grade classification using if-elif-else ──────────────────────────
score = 73   # student's exam score

if score >= 90:               # 90 and above
    grade = 'A (Excellent)'
elif score >= 80:             # 80 to 89
    grade = 'B (Good)'
elif score >= 70:             # 70 to 79
    grade = 'C (Average)'
elif score >= 60:             # 60 to 69
    grade = 'D (Below Average)'
else:                         # below 60
    grade = 'F (Fail)'

print(f'Score: {score} → Grade: {grade}')
```

```python
# ── Logical operators: and / or / not ────────────────────────────────
x = 15   # test number

# 'and' – both conditions must be True
if x > 10 and x < 20:
    print(f'{x} is between 10 and 20')   # True and True → prints

# 'or' – at least one condition must be True
if x < 5 or x > 12:
    print(f'{x} is outside the range 5–12')  # False or True → prints

# 'not' – inverts the condition
if not x == 10:
    print(f'{x} is not equal to 10')    # not False → True → prints
```

```python
# ── Checking divisibility (modulo + conditionals) ────────────────────
n = 36   # number to test

if n % 2 == 0:          # remainder when divided by 2 is 0?
    print(f'{n} is Even')
else:
    print(f'{n} is Odd')

if n % 3 == 0 and n % 5 == 0:   # divisible by both 3 and 5?
    print(f'{n} is divisible by both 3 and 5')
elif n % 3 == 0:
    print(f'{n} is divisible by 3 only')
elif n % 5 == 0:
    print(f'{n} is divisible by 5 only')
```

06 – Loops & Mathematical Iteration

```python
# ── for loop with range ───────────────────────────────────────────────
# range(1, 6) generates: 1, 2, 3, 4, 5
print('Counting from 1 to 5:')
for i in range(1, 6):      # i takes values 1, 2, 3, 4, 5 in order
    print(f'  i = {i}')
```

```python
# ── Sum of first n natural numbers: S = n(n+1)/2 ─────────────────────
n = 10        # we want the sum of 1 + 2 + ... + 10
total = 0     # start with sum = 0

for i in range(1, n + 1):   # i goes 1, 2, 3, ..., 10
    total += i               # add i to running total (total = total + i)

# Verify using the mathematical formula: n(n+1)/2
formula_result = n * (n + 1) // 2   # integer division

print(f'Sum (loop)    : {total}')          # 55
print(f'Sum (formula) : {formula_result}') # 55 — should match!
```

```python

# ── Factorial: n! = 1 × 2 × 3 × ... × n ─────────────────────────────
n = 6           # compute 6!
factorial = 1   # start with product = 1 (not 0, as we are multiplying)

for i in range(1, n + 1):   # i goes 1, 2, 3, 4, 5, 6
    factorial *= i           # multiply accumulator by i
    print(f'  Step {i}: {i}! so far = {factorial}')

print(f'\n6! = {factorial}')  # 720
```

```python
# ── while loop: digits of Fibonacci sequence ─────────────────────────
# Fibonacci: each number = sum of the two before it → 0, 1, 1, 2, 3, 5, 8, ...
a, b = 0, 1       # initialise first two Fibonacci numbers
limit = 100       # print all Fibonacci numbers below 100

print('Fibonacci numbers below', limit, ':')
while a < limit:           # keep looping while a is less than limit
    print(a, end='  ')     # print current value (end=' ' keeps on same line)
    a, b = b, a + b        # update: next = current + previous (simultaneous swap)
```

```python
# ── Nested loop: multiplication table ────────────────────────────────
print('Multiplication Table (1–5)\n')
print('   ', end='')              # header spacing
for col in range(1, 6):
    print(f'{col:4}', end='')     # column headers
print()                           # newline after header
print('─' * 25)

for row in range(1, 6):           # outer loop: rows 1 to 5
    print(f'{row:2} |', end='')   # row label
    for col in range(1, 6):       # inner loop: columns 1 to 5
        print(f'{row*col:4}', end='')  # compute and print product
    print()                       # newline at end of each row
```
