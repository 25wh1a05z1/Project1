```python
import numpy as np
```

```python
math=np.array([89,83,95])
science=np.array([91,93,98])
english=np.array([92,87,96])
print("Math score: ",math)
print("Science score: ",science)
print("English score: ",english)
```

```python
from numpy.linalg import norm
def calculate_cosine_similarity(v1,v2):
  dot_product=np.dot(v1,v2)
  magnitude=norm(v1)*norm(v2)
  if magnitude==0:
    return 0
  return dot_product/magnitude
cosine_math_science=calculate_cosine_similarity(math,science)
print(f"Cosine similarity between math and science: {cosine_math_science: .4f}")
cosine_math_english=calculate_cosine_similarity(math,english)
print(f"Cosine similarity between math and english: {cosine_math_english: .4f}")
cosine_science_english=calculate_cosine_similarity(science,english)
print(f"Cosine similarity between science and english: {cosine_science_english: .4f}")
```

```python
A= np.array([
    [89,91,92],
    [83,93,87],
    [95,98,96]
])
B= np.array([50,70,45])
print("Coefficient matrix (A): ")
print(A)
print("Coefficient matrix (B): ")
print(B)
```

```python
solution = np.linalg.solve(A,B)
price_A,price_B,price_C=solution
print(f"The price of A: ${price_A: .2f}")
print(f"The price of B: ${price_B: .2f}")
print(f"The price of C: ${price_C: .2f}")
```

🧪 LAB EXERCISE 2 — Central tendency
For the tip column:

Print its mean, median and mode.

Is it right-skewed? Check whether mean > median.

Plot a histogram of tip with vertical lines at the mean and median.

```python
# LAB EXERCISE 2
# Central tendency of the tip column

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset("tips")

# Calculate mean, median and mode
mean_tip = df["tip"].mean()
median_tip = df["tip"].median()
mode_tip = df["tip"].mode()[0]

print("Mean:", mean_tip)
print("Median:", median_tip)
print("Mode:", mode_tip)

# Check whether the data is right-skewed
if mean_tip > median_tip:
    print("\nMean > Median, so tip is right-skewed.")
else:
    print("\nMean is not greater than median, so it is not right-skewed based on this check.")

# Plot histogram
plt.figure(figsize=(8, 5))

plt.hist(df["tip"], bins=20, color="skyblue", edgecolor="black")

# Vertical line for mean
plt.axvline(mean_tip, color="red", linestyle="--", linewidth=2,
            label=f"Mean = {mean_tip:.2f}")

# Vertical line for median
plt.axvline(median_tip, color="green", linestyle="-", linewidth=2,
            label=f"Median = {median_tip:.2f}")

plt.title("Histogram of Tip")
plt.xlabel("Tip")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

🧪 LAB EXERCISE 3 — Spread
For the tip column:

Print the range, variance and standard deviation.

Compute Q1, Q3 and the IQR.

In a comment, say which measure of spread is most robust to outliers and why.

```python
# LAB EXERCISE 3
# Measures of spread for the tip column

import pandas as pd
import seaborn as sns

# Load dataset
df = sns.load_dataset("tips")

# Calculate range
tip_range = df["tip"].max() - df["tip"].min()

# Calculate variance
tip_variance = df["tip"].var()

# Calculate standard deviation
tip_std = df["tip"].std()

# Calculate Q1 and Q3
Q1 = df["tip"].quantile(0.25)
Q3 = df["tip"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

print("Range:", tip_range)
print("Variance:", tip_variance)
print("Standard Deviation:", tip_std)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

# IQR is the most robust measure of spread because
# it is less affected by extreme values/outliers than range,
# variance, and standard deviation.
```

🧪 LAB EXERCISE 4 — Five-number summary & box plot
For the tip column:

Print its five-number summary (min, 25%, 50%, 75%, max).

Print its skewness and say (in a comment) which way it leans.

Draw a box plot of tip.

```python
# LAB EXERCISE 4
# Five-number summary, skewness and box plot

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset("tips")

# Five-number summary
five_number_summary = df["tip"].describe()[["min", "25%", "50%", "75%", "max"]]

print("Five-number summary:")
print(five_number_summary)

# Calculate skewness
skewness = df["tip"].skew()

print("\nSkewness of tip:", skewness)

# Interpret skewness
if skewness > 0:
    print("The distribution leans to the right (positive/right skew).")
elif skewness < 0:
    print("The distribution leans to the left (negative/left skew).")
else:
    print("The distribution is approximately symmetric.")

# Box plot
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["tip"], color="lightgreen")

plt.title("Box Plot of Tip")
plt.xlabel("Tip")

plt.show()
```

🧪 LAB EXERCISE 5 — Correlation & full summary

Build the correlation matrix and print it (rounded to 2 decimals).

Draw a correlation heatmap.

Run df.describe() and, in a comment, write 3 insights about the dataset (e.g. which column is most spread out, which pair is most correlated, any skew you notice).

```python
# LAB EXERCISE 5
# Correlation matrix, heatmap and descriptive statistics

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset("tips")

# Select numerical columns
numerical_df = df.select_dtypes(include="number")

# Create correlation matrix
correlation_matrix = numerical_df.corr()

print("Correlation Matrix:")
print(correlation_matrix.round(2))

# Draw correlation heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()

# Full descriptive summary
print("\nDescriptive Statistics:")
print(df.describe())

```
