**Task 1:** Python Fundamentals, Lists, Tuples, Sets, Dictionaries, Indexing & Slicing

**1.1:-** Generate a list of 100 simulated sensor readings.
Each reading should be a tuple: (timestamp_str, device_id, device_type, location, value).
*   timestamp_str: Format as 'YYYY-MM-DD HH:MM:SS'
*   device_id: A unique ID for each reading (e.g., 'DEV001', 'DEV002', etc., up to 100 unique IDs).
*   device_type: Randomly chosen from DEVICE_TYPES.
*   location: Randomly chosen from LOCATIONS.
*   value: Random float within READING_RANGES for its device_type (round to 2 decimal places).
*   Store these in a list called 'raw_sensor_data'.







```python
import datetime
import random

# Constants for Task 1.1
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['RoomA', 'RoomB', 'RoomC', 'Hallway', 'Laboratory']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 70.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100.0, 1000.0)
}

raw_sensor_data = []
start_time = datetime.datetime(2023, 1, 1, 8, 0, 0)

for i in range(100):
    # Timestamp
    timestamp = start_time + datetime.timedelta(minutes=i)
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    # Device ID
    device_id = f'DEV{i+1:03d}'

    # Device Type
    device_type = random.choice(DEVICE_TYPES)

    # Location
    location = random.choice(LOCATIONS)

    # Value
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    raw_sensor_data.append((timestamp_str, device_id, device_type, location, value))

print(f"Generated {len(raw_sensor_data)} raw sensor readings.")
print("First 5 readings:")
for reading in raw_sensor_data[:5]:
    print(reading)
```

**1.2:-** Create a list named 'temperature_readings'
This list will contain only the readings where 'device_type' is 'Temperature_Sensor'. Each item in 'temperature_readings' should be a dictionary with keys: 'timestamp', 'device_id', 'location', 'temperature_celsius'.

```python
temperature_readings = []

for reading in raw_sensor_data:
    timestamp, device_id, device_type, location, value = reading
    if device_type == 'Temperature_Sensor':
        temperature_readings.append({
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        })

print(f"Generated {len(temperature_readings)} temperature readings.")
print("First 5 temperature readings (dictionaries):")
for reading_dict in temperature_readings[:5]:
    print(reading_dict)
```

**1.3:-** Create a set named 'unique_locations'
This set will contain all unique locations from 'raw_sensor_data'.

```python
unique_locations = set()

for reading in raw_sensor_data:
    # Location is at index 3 in the tuple
    unique_locations.add(reading[3])

print("Unique locations:", unique_locations)
```

**1.4:-** Create a dictionary named 'device_stats'
Keys will be 'device_type' and values will be a tuple: (min_value, max_value) observed for that device type across all readings. Initialize with sensible large/small values.

```python
device_stats = {}

# Initialize device_stats with very large/small values
for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf')) # (min_val, max_val)

for reading in raw_sensor_data:
    _, _, device_type, _, value = reading
    current_min, current_max = device_stats[device_type]
    device_stats[device_type] = (
        min(current_min, value),
        max(current_max, value)
    )

print("Device statistics (min/max values per device type):")
for device_type, stats in device_stats.items():
    print(f"  {device_type}: Min={stats[0]}, Max={stats[1]}")
```

**Task 2:** Writing and Using Functions in Python

**2.1 :** get_readings_by_location(readings_list, target_location)
This function takes a list of sensor readings (tuples) and a target location string. It returns a list of all readings for that specific location.

```python
raw_sensor_data = [
    ("2023-10-27 10:00:00", "device001", "RoomA", 22.5),
    ("2023-10-27 10:01:00", "device002", "RoomB", 24.1),
    ("2023-10-27 10:02:00", "device001", "RoomA", 22.7),
    ("2023-10-27 10:03:00", "device003", "RoomC", 23.0),
    ("2023-10-27 10:04:00", "device002", "RoomB", 24.3),
]

def get_readings_by_location(readings_list, target_location):
    """
    Filters a list of sensor readings by a target location.

    Args:
        readings_list (list): A list of tuples, where each tuple represents a sensor reading
                              (timestamp, device_id, location, temperature).
        target_location (str): The location to filter by.

    Returns:
        list: A list of readings (tuples) for the specified target location.
    """
    filtered_readings = []
    for reading in readings_list:
        # Assuming location is the 3rd element (index 2) in the tuple
        if reading[2] == target_location:
            filtered_readings.append(reading)
    return filtered_readings

# Example Usage for 2.1:
room_a_readings = get_readings_by_location(raw_sensor_data, "RoomA")
print("Readings for RoomA:", room_a_readings)

room_b_readings = get_readings_by_location(raw_sensor_data, "RoomB")
print("Readings for RoomB:", room_b_readings)
```

**2.2**:- calculate_average_temperature(temp_data_list)
This function takes a list of temperature reading dictionaries. It calculates and returns the average temperature. It should return 0.0 if the list is empty.

```python
temperature_readings = [
    {"timestamp": "2023-10-27 10:00:00", "device_id": "device001", "location": "RoomA", "temperature_celsius": 22.5},
    {"timestamp": "2023-10-27 10:01:00", "device_id": "device002", "location": "RoomB", "temperature_celsius": 24.1},
    {"timestamp": "2023-10-27 10:02:00", "device_id": "device001", "location": "RoomA", "temperature_celsius": 22.7},
    {"timestamp": "2023-10-27 10:03:00", "device_id": "device003", "location": "RoomC", "temperature_celsius": 23.0},
    {"timestamp": "2023-10-27 10:04:00", "device_id": "device002", "location": "RoomB", "temperature_celsius": 24.3},
]

def calculate_average_temperature(temp_data_list):
    """
    Calculates the average temperature from a list of temperature reading dictionaries.

    Args:
        temp_data_list (list): A list of dictionaries, where each dictionary contains
                               'temperature_celsius' key.

    Returns:
        float: The average temperature, or 0.0 if the list is empty.
    """
    if not temp_data_list:
        return 0.0
    total_temperature = sum(d['temperature_celsius'] for d in temp_data_list)
    return total_temperature / len(temp_data_list)

# Example Usage for 2.2:
average_temp = calculate_average_temperature(temperature_readings)
print("Average Temperature:", average_temp)

empty_list_avg = calculate_average_temperature([])
print("Average Temperature (empty list):", empty_list_avg)
```

**2.3**:- generate_report_line(reading_dict)
This function takes a single temperature reading dictionary. It returns a formatted string like: "[{timestamp}] Device {device_id} at {location}: {temperature_celsius}°C"

```python
def generate_report_line(reading_dict):
    """
    Generates a formatted report string from a single temperature reading dictionary.

    Args:
        reading_dict (dict): A dictionary containing 'timestamp', 'device_id',
                             'location', and 'temperature_celsius' keys.

    Returns:
        str: A formatted string representing the report line.
    """
    timestamp = reading_dict.get("timestamp", "N/A")
    device_id = reading_dict.get("device_id", "N/A")
    location = reading_dict.get("location", "N/A")
    temperature_celsius = reading_dict.get("temperature_celsius", "N/A")
    return f"[{timestamp}] Device {device_id} at {location}: {temperature_celsius}°C"

# Example Usage for 2.3:
sample_reading = {
    "timestamp": "2023-10-27 10:00:00",
    "device_id": "device001",
    "location": "RoomA",
    "temperature_celsius": 22.5
}
report_line = generate_report_line(sample_reading)
print("Report Line:", report_line)

sample_reading_2 = {
    "timestamp": "2023-10-27 10:04:00",
    "device_id": "device002",
    "location": "RoomB",
    "temperature_celsius": 24.3
}
report_line_2 = generate_report_line(sample_reading_2)
print("Report Line:", report_line_2)
```

**3.2:-**  Calculate the duration between the first and last timestamp
This task calculates the duration between the first and last timestamp in raw_sensor_data, converting string timestamps to datetime objects and printing the duration in days, hours, and minutes.

```python
import datetime

# raw_sensor_data is available from previous cells.
# raw_sensor_data = [
#     ("2023-10-27 10:00:00", "device001", "RoomA", 22.5),
#     ("2023-10-27 10:01:00", "device002", "RoomB", 24.1),
#     ("2023-10-27 10:02:00", "device001", "RoomA", 22.7),
#     ("2023-10-27 10:03:00", "device003", "RoomC", 23.0),
#     ("2023-10-27 10:04:00", "device002", "RoomB", 24.3),
# ]

if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    # Convert string to datetime object
    # The format string must exactly match the timestamp string
    date_format = '%Y-%m-%d %H:%M:%S'
    first_dt = datetime.datetime.strptime(first_timestamp_str, date_format)
    last_dt = datetime.datetime.strptime(last_timestamp_str, date_format)

    duration = last_dt - first_dt

    days = duration.days
    seconds = duration.total_seconds()
    hours = int(seconds // 3600) % 24 # Calculate remaining hours after full days
    minutes = int((seconds % 3600) // 60) # Calculate remaining minutes after full hours

    print(f"\nDuration of data collection: {days} days, {hours} hours, {minutes} minutes.")
else:
    print("\nNot enough data to calculate duration.")
```

This task writes all temperature_readings to a file named 'sensor_reports/temperature_report.txt'. Each line in the file is generated using the generate_report_line function.

```python
read_report_lines = []
try:
    with open(report_filename, 'r') as f:
        read_report_lines = f.readlines()
    print(f"\nSuccessfully read {len(read_report_lines)} lines from '{report_filename}'.")
    print("First 3 lines of the report:")
    for line in read_report_lines[:3]:
        print(line.strip())
except FileNotFoundError:
    print(f"\nError: File '{report_filename}' not found.")
except IOError as e:
    print(f"\nError reading file '{report_filename}': {e}")
```
